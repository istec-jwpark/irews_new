```python
import hashlib
```

```python
pwd = "test\r\n"
hashlib.sha256(pwd.encode('utf-8')).hexdigest()
```

```python
text = "test\n"
# SHA-256 해싱 실행
hash_result = hashlib.sha256(text.encode('utf-8')).hexdigest()
print(hash_result)
```
