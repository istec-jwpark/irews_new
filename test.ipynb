```python
import hashlib
def test_various_hashes(password: str):
    # 1. 일반 SHA-256 (사용자가 확인한 값: 9f86d...)
    print(f"Single: {hashlib.sha256(password.encode()).hexdigest()}")
    
    # 2. Double SHA-256 (두 번 해싱)
    first_round = hashlib.sha256(password.encode()).digest()
    second_round = hashlib.sha256(first_round).hexdigest()
    print(f"Double: {second_round}")
    
    # 3. Uppercase 변환 후 해싱
    upper_hash = hashlib.sha256(password.upper().encode()).hexdigest()
    print(f"Upper:  {upper_hash}")
```

```python
test_various_hashes("test")
```

```python
def get_double_sha256(password: str) -> str:
    # 1차 해싱
    first_round = hashlib.sha256(password.encode()).hexdigest()
    # 2차 해싱 (자바스크립트의 SHA256(SHA256(value))와 동일)
    second_round = hashlib.sha256(first_round.encode()).hexdigest()
    return second_round
```
